```python
def reverse2(F_str: str):
    F = isl.Map(F_str)

    rev = F.reverse().coalesce()

    # Full identity map on same space
    ident = isl.Map.identity(rev.get_space())

    # Domain where rev is defined
    dom = rev.domain()

    # Complement of domain
    not_dom = ident.domain().subtract(dom)

    # Identity only outside rev's domain
    dom_ident = ident.intersect_domain(dom)
    
    ident_elsewhere = ident.intersect_domain(not_dom)

    # Piecewise map:
    #   rev on dom
    #   identity elsewhere
    res = rev.intersect_domain(dom).union(ident_elsewhere).coalesce()

    return res
```
