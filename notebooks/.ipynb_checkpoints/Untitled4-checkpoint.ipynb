```python
from sympy import Rational

def exponential_to_1_through_3_exact(x0, h, y0, y1, y2, Z):
    y0 = Rational(y0)
    y1 = Rational(y1)
    y2 = Rational(y2)

    ah = ((1 - y0) / (1 - y1)).simplify()
    a = ah**(Rational(1, h))

    if ((1 - y2) - a**(Z - (x0 + 2*h))).simplify() != 0:
        raise ValueError("Points do not fit 1 - a^(Z-x) exactly")

    return a


# known constant
Z = 5

# sample points
x0 = 0
h = 1

y0 = -31
y1 = -15
y2 = -7

a = exponential_to_1_through_3_exact(x0, h, y0, y1, y2, Z)

print(a)
```
