```python
import islpy as isl

ctx = isl.DEFAULT_CONTEXT

odd = isl.Set.read_from_str(
    ctx,
    "{ [x] : exists (a : x = 2*a + 1) }"
)

even = isl.Set.read_from_str(
    ctx,
    "{ [x] : exists (a : x = 2*a) }"
)

f_odd = isl.PwAff.read_from_str(ctx, "{ [x] -> 0 }").intersect_domain(odd)
f_even = isl.PwAff.read_from_str(ctx, "{ [x] -> 1 }").intersect_domain(even)

f = f_odd.union_add(f_even)

print(f)
```
