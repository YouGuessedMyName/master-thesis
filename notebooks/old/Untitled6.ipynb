```python
import islpy as isl

ctx = isl.DEFAULT_CONTEXT

# y <= x !

x = isl.PwAff("{ [c, g] -> [(1)] : g = 1 and 0 <= c <= 9; "
              "[c, g] -> [(3178624404039435 - 239339748956940c)/4503599627370496] : g = 0 and 2 <= c <= 3; "
              "[c, g] -> [(3403190094418787 - 295481171551778c)/4503599627370496] : g = 0 and 4 <= c <= 5; "
              "[c, g] -> [(3819052484010179 - 364791569817010c)/4503599627370496] : g = 0 and 6 <= c <= 7; "
              "[c, g] -> [(6175350598871596 - 387730393310243c)/9007199254740992] : g = 0 and 0 <= c <= 1; "
              "[c, g] -> [(9007199254740990 - 900719925474099c)/9007199254740992] : g = 0 and 8 <= c <= 9 }")

y = isl.PwAff("{ [c, g] -> [(1)] : g = 1 and 0 <= c <= 9; "
              "[c, g] -> [(1)/10] : c = 9 and g = 0; "
              "[c, g] -> [(19)/100] : c = 8 and g = 0; "
              "[c, g] -> [(271)/1000] : c = 7 and g = 0; "
              "[c, g] -> [(3439)/10000] : g = 0 and 0 <= c <= 6 }")

# point [c,g] = [9,0]
space = isl.Space.create_from_names(ctx, set=["c", "g"])

p = isl.Point.zero(space)
p = p.set_coordinate_val(isl.dim_type.set, 0, isl.Val.int_from_si(ctx, 9))
p = p.set_coordinate_val(isl.dim_type.set, 1, isl.Val.int_from_si(ctx, 0))

print("x(9,0) =", x.eval(p)) # but this evaluates slightly too high :D
print("y(9,0) =", y.eval(p))
```
