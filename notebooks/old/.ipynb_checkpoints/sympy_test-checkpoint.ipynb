```python
from sympy import *
```

```python
c,g = symbols('c g')

pw = Piecewise((0, Eq(g, 0) & (c >= 0) & (c <= 15)), (1, Eq(g, 1) & (c >= 0) & (c <= 15)))
print(pw)
substed = pw.subs({c: 2, g: 1})
print(substed)
```

```python
import islpy as isl

pw = isl.PwQPolynomial("{ [c, g] -> 1 : g = 1 and 0 <= c <= 15; [c, g] -> ((9/10 - 3/50 * c) * c) : g = 0 and 0 <= c <= 15 }")
```

```python
for dom, expr in pw.get_pieces():
    print(dom, expr)
    x = expr.as_aff()
    print(x)
```
